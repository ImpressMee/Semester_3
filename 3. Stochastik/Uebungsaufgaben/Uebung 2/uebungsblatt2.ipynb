```python
import math

# Alternativ ginge auch math.comb(n,k)

def kombiOhneWiederholung(n,k):
    return math.factorial(n) / (math.factorial(k) * math.factorial(n - k))
```

```python
# Aufgabe 2.6
'''
ohne zurücklegen, es werden keine Fragen doppelt gestellt.

Kombination ohne wiederholung: 
- Binominalkoeffizient: (n k) = n! / (k! * (n-k)!)
- n: Anzahl aller Elemente
- k: Anzahl der auszuwählenden Elemente
'''

# Aufgabe 2.6.1

'''
Felix: nur 1 Frage
10 Fragen, 2 davon werden ausgewählt
(10 2) = 45
9 Mögliche zweite Fragen, Eine davon ist von Felix
(9 1) = 9 

P(1 Frage) = günstige Fälle/alle Fälle = 9/45 = 0,2
'''
alle_faelle = kombiOhneWiederholung(10, 2)
guenstige_faelle = kombiOhneWiederholung(9,1)

P = guenstige_faelle / alle_faelle
print(f"Aufgabe 2.6.1\nP(Felix bekommt seine einzige Frage) = {P}")

# Aufgabe 2.6.2
'''
Alexandra: 5 Fragen
P(Frage kommt nicht dran) = 5/10 * 4/9 = 2/9
P(Mind 1 Frage) = 1 - 2/9 = 7/9 ~ 77,8%
'''
alle_faelle = kombiOhneWiederholung(10, 2)
ungünstige_faelle = kombiOhneWiederholung(5, 2) # 10 Fragen, kommt keine einzige. 5 unvorbereitete, 2 gezogene
guenstige_faelle = alle_faelle - ungünstige_faelle

wahrscheinlichkeit = guenstige_faelle / alle_faelle

print(f"\nAufgabe 2.6.2\nP(Mindestens 1 Frage) = {wahrscheinlichkeit:.3f}")


```

```python
import math
# alternativ: math.comb(n + k - 1, k)
def kombiMitWiederholung(n, k):
    return math.factorial(n + k - 1) / (math.factorial(k) * math.factorial(n - 1))

def variation(n,k):
    return math.factorial(n)/math.factorial(n-k)
```

```python
# Aufgabe 2.8 Nutzernamen

'''
n: 26 kleinbuchstaben
Typ 1: k = 3 verschiedene buchstaben (ohne zurück legen)
Typ 2: k = 4 buchstaben (mit zurücklegen)
Typ 3: k = 5 ODER 6 Kleinbuchstaben (ohne zurück legen)

Typ 
'''
# Aufgabe 2.8.1

T1 = kombiOhneWiederholung(26,3)
T2 = kombiMitWiederholung(26, 4)
T3 = kombiOhneWiederholung(26,5) + kombiOhneWiederholung(26,6)

print(f"Typ 1 = {T1}\nTyp 2 = {T2}\nTyp 3 = {T3}")
```

```python
# Aufgabe 2.8.2

'''
a)
- Zufällig ausgewählter typ 1
- Anz aller Kombinationen
- Anz aller Kombinationen ohne x
- berechnung von Alle kombinationen mit mit x
- berechnung der warscheinlichkeit: Komb mit x / alle Kombinationen
'''
a = (T1 - kombiOhneWiederholung(25,3)) / T1

'''
b)
0%.

Typ 1 definiert durch 3 VERSCHIEDENE buchstaben. Demensprechend kann 'xxx' garnicht vorkommen
'''
b = 0

'''
c)
- Typ 2
- x kommt garnicht dran
- alle Kombinationen mit wiederholung
- alle kombinationen ohne x
- warscheinlichkeit =  alle ohne x / alle mit wiederholung
'''
c = kombiMitWiederholung(25,4) / T2

'''
d)
Typ 3
- entweder 5 oder 6 buchstaben, nicht wiederholt
- abcde oder uvwxyz, reihenfolge wichtig!
- deswegen beide variationen für 5 und 6 buchstaben berechnen
- 2 Günstige Fälle
- günstige Fälle / Mögliche Kombinationen (Varationen)

'''

d = 2 / (variation(26,5) + variation(26,6))
print(f"a) {a:.3f}")
print(f"b) {b}")
print(f"c) {c:.3f}")
print(f"d) {d:.2e}")
```
