You need to use numpy in order to read data from files or use a list, tuples, matrices oder funtion which return numpy arrays, such as arange, linespace, and more.

```python
# a vector: the argument to the array funtion is a Python list
v = np.array([1,2,3,4])

v
```

```python
# a matrix: the argument to the array function is a nested python list
M = np.array([[1,2],[3,4]])

M
```
