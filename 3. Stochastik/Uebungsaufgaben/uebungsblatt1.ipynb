```python
import math
# Aufgabe 1.5
# Aufgabe 1.5.1
urliste = [25, 10, 30, 25, 35, 25]
sortierteliste = sorted(urliste)
arithmetisches_mittel = np.mean(sortierteliste)
der_median = np.median(sortierteliste)
q1 = np.percentile(sortierteliste, 75)

print(f"a) arithmetisches mittel: {arithmetisches_mittel}")
print(f"b) der median ist: {der_median}")
print(f"c) Das 75% Quantil ist: {q1}")

# d) Empirische Standartabweichung
n = len(urliste)

x = 0
print("d) ----------------------")

for werte in range(n):
    # Wenn es nicht das letzte Element ist → Plus drucken
    if werte < n - 1:
        print(f"({sortierteliste[werte]} - {arithmetisches_mittel})² +", end=" ")
    else:
        print(f"({sortierteliste[werte]} - {arithmetisches_mittel})²", end="")

    x += (sortierteliste[werte] - arithmetisches_mittel) ** 2

print(f"\nDas Ergebnis der Summe ist {x}")
s2 = x/n-1 # empirische Varianz
s = math.sqrt(s2) # empirische Standartabweichung

print("-------------------------")
print(f"Empirische Varianz: s2 = ({x} / {n}-1) = {s2}")
print(f"Empirische Standartabweichung:s = sqrt({s2}) = {s}")


```

```python
# Teilaufgabe 1.10.1
# Quantitativ Diskret, denn die Anzahl der Räder sind messbare Zahlen und Diskret da zwischenwerte nicht möglich sind

# Teilaufgabe 1.10.2
v_hersteller = np.array([4.2, 3.9, 4.3, 4.1, 4.1, 3.7, 4.3])
v_test = np.array([4.8, 4.1, 4.2, 4.7, 3.9, 5.0])
# a)
arithmetisches_mittel = np.mean(v_hersteller)

# b)
q25 = np.percentile(sorted(v_hersteller), 25)
q50 = np.percentile(sorted(v_hersteller), 50)
q75 = np.percentile(sorted(v_hersteller), 75)

# c)
q10 = np.percentile(sorted(v_hersteller), 10)

# d)
spannweite = max(v_hersteller) - min(v_hersteller)

# e)

varianz = 0 # varianz berechnung
for i in range(len(v_hersteller)):
    varianz += (v_hersteller[i] - arithmetisches_mittel)**2

varianz = varianz / (len(v_hersteller) - 1)

s = math.sqrt(varianz)  # Sigma standartabweichung
                        # Sigma weil wir die Gesamtheit betrachten

print("Sortedarray: ", sorted(v_hersteller))        
print(f"a)\narithmetisches mittel: {arithmetisches_mittel}")
print(f"b)\nQuantil 25%: {q25}, Quantil 50%: {q50}, Quantil 75%: {q75}")
print(f"c)\nQuantil 10%: {q10}")
print(f"d)\nSpannweite: {spannweite}")
print(f"e)\nEmpirische Standartabweichung: {s}")
```

```python
# Aufgabe 1.12 a)

# Werte

ausbildung_x = np.array([9, 13, 15, 18, 20])
jahresgehalt_y = np.array([18, 37, 61, 125, 59])

# Mittelwerte

mean_x = np.mean(ausbildung_x)
mean_y = np.mean(jahresgehalt_y)

# Kovarianz
summe = 0
for i in range(len(ausbildung_x)):
    summe += (ausbildung_x[i] - mean_x) * (jahresgehalt_y[i] - mean_y)
s_xy = summe / (len(ausbildung_x) - 1)

# Standardabweichungen (empirisch)
s_x = np.sqrt(np.sum((ausbildung_x - mean_x)**2) / (len(ausbildung_x) - 1))
s_y = np.sqrt(np.sum((jahresgehalt_y - mean_y)**2) / (len(jahresgehalt_y) - 1))

# Korrelationskoeffizient
r = s_xy / (s_x * s_y)

print(f"Empirische Kovarianz s_xy: {s_xy:.2f}")
print(f"Empirischer Korrelationskoeffizient r: {r:.4f}")
```

# Lineare Regression
Sehr wichtig für KI!



oder anders dargestellt:



- E = Gesamtfehler
- n = Anzahl der Messpunkte (Datenpunkte)
- x_i = Eingabewert
- y_i = Zielwert
- a (k) = steigung der Geraden
- b (d) = Achsenabschnitt (Schnittpunkt mit der y-Achse)

```python
import matplotlib.pyplot as plt

# Regressionsparameter
k = s_xy / (s_x**2)
d = mean_y - k * mean_x

# Vorhersagewerte (Gerade)
y_pred = k * ausbildung_x + d

# Gesamtfehler (Summe der Quadrate)
E = np.sum((jahresgehalt_y - y_pred)**2)
print(f"Gesamtfehler E = {E:.2f}")

# Plot
plt.figure(figsize=(6, 4))
plt.scatter(ausbildung_x, jahresgehalt_y, color='blue', label='Datenpunkte')
plt.plot(ausbildung_x, y_pred, color='red', label='Regressionsgerade')
plt.xlabel("Ausbildungsjahre")
plt.ylabel("Jahresgehalt (in Tsd. €)")
plt.title("Lineare Regression: Ausbildungsdauer vs. Jahresgehalt")
plt.legend()
plt.grid(True)
plt.show()
plt.show()
```
