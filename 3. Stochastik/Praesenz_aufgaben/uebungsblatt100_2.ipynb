```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Aufgabe 1.2.1
werte = pd.read_csv("ninjagokarten.csv", delimiter=";")
print(werte)
```

```python
# aufgabe 100.2.2
# Mittelwert
gelb_mean = np.mean(werte["rot"])
gruen_mean = np.mean(werte["gruen"])
rot_mean = np.mean(werte["rot"])
blau_mean = np.mean(werte["blau"])

print(f"Rot mean: {rot_mean:.3f}")
print(f"Gruen mean: {gruen_mean:.3f}")
print(f"Blau mean: {blau_mean:.3f}")
print(f"Gelb mean: {gelb_mean:.3f}")

# Standartabweichung
gelb_std = np.std(werte["gelb"], ddof=1)
gruen_std = np.std(werte["gruen"], ddof=1)
rot_std = np.std(werte["rot"], ddof=1)
blau_std = np.std(werte["blau"], ddof=1)
print("-----------------------------------------")
print(f"Standartabweichung Gelb: {gelb_std:.3f}")
print(f"Standartabweichung Grün: {gruen_std:.3f}")
print(f"Standartabweichung Rot: {rot_std:.3f}")
print(f"Standartabweichung Blau: {blau_std:.3f}")
```

```python

from matplotlib.pyplot import bar

df = werte.drop("Name", axis=1)
plt.figure()
plt.ylabel("Werte")
plt.xlabel("Eigenschaften")
plt.bar(bar(['gelb', 'rot', 'blau', 'gruen'], [1,2,3,4]),werte)
```

```python
# Aufgabe 100.2.4
#values, counts = np.unique(werte["rot"], return_counts=True)
plt.boxplot(df, tick_labels=["gelb", "rot", "blau", "gruen"])
plt.xlabel("Eigenschaften")
plt.ylabel("Werte")
```
