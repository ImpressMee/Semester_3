```python
# Teilaufgabe 6.1.1
'''
Die Dichte ist die Ableitung von F(x)

X ist gleichverteilt [2,4]

a)  f(x) = F'(x) = 1/2: für 2 < x < 4 ; 0: sonstiges
b)  P(X < 3) nicht größer 3
    P(x < 3) = F(3) = 3/2 - 1 = 1/2
c)  P((5/2) <= X <= (7/2))
    F(7/2) - F(5/2) = ((7/4) - 1) - ((5/4) - 1) = 1/2
'''
import numpy as np

# Parameter der Gleichverteilung
a, b = 2, 4

# Dichtefunktion
def f(x):
    return np.where((x > a) & (x < b), 0.5, 0.0)

# Verteilungsfunktion
def F(x):
    return np.where(
        x <= a, 0,
        np.where(x < b, x/2 - 1, 1)
    )

# a) Dichte prüfen (Beispielwerte)
x_test = np.array([1, 2.5, 4.5])
print("f(x):", f(x_test))

# b) P(X < 3)
p_b = F(3)
print("P(X < 3) =", p_b)

# c) P(5/2 ≤ X ≤ 7/2)
p_c = F(7/2) - F(5/2)
print("P(5/2 ≤ X ≤ 7/2) =", p_c)
```

```python
# Teilaufgabe 6.1.2

import sympy as sp

x = sp.symbols('x')
f = sp.Rational(1, 2)

# Erwartungswert
E = sp.integrate(x * f, (x, 2, 4))

# Varianz (nach Definition!)
Var = sp.integrate((x - E)**2 * f, (x, 2, 4))

Std = sp.sqrt(Var)

E, Var, Std

```
