### Als aller erstes werden die Daten eingelesen:

1. **Import glob:**  
   sorgt dafür, dass man eine Art Wildcard verwenden kann.  
   Dadurch können alle passenden Dateien automatisch gefunden und in `dateien` gespeichert werden.

2. **Converters:**  
   Diese Variable speichert zwei Lambda-Funktionen, die in den Spalten das **Komma (`,`)** durch einen **Punkt (`.`)** ersetzen.  
   So werden die Werte korrekt als `float` eingelesen.

3. **Einlesen der Dateien:**  
   Alle gefundenen Dateien werden mit einer Schleife (bzw. List Comprehension) eingelesen und in `daten` gespeichert.

4. **DINA4-Messungen:**  
   Die zusätzlichen DINA4-Messungen werden separat eingelesen und in eigenen Variablen gespeichert.


```python
import numpy as np
import matplotlib.pyplot as plt
import glob

# Alle Versuch-Dateien finden
dateien = sorted(glob.glob("messungen/Versuch1_*cm.csv")) # * ist eine Wildcard. Alles was in 
                                                          # dieses Muster passt, wird in Dateien eingetragen
print(dateien)

# Komma in Punkt umwandeln
converters = {
    0: lambda s: float(s.replace(',', '.')), # Das ist für die Zeit
    1: lambda s: float(s.replace(',', '.')) # Das ist für den Volt wert
}

m1 = np.genfromtxt("messungen/Versuch1_12cm.csv", delimiter=";", skip_header=1000, converters=converters)

# Alle Dateien einlesen
daten = [np.genfromtxt(filename, delimiter=";", skip_header=1000, converters=converters) for filename in dateien]

print(daten)

# DINA4-Messungen
messungenDINA4_breite = np.genfromtxt("messungen/Versuch1_DINA4_BREITE.csv", delimiter=";", skip_header=1000, converters=converters)
messungenDINA4_laenge = np.genfromtxt("messungen/Versuch1_DINA4_LAENGE.csv", delimiter=";", skip_header=1000, converters=converters)
```

Erstellung eines Plots um die Daten zu visualisieren

```python
# Jede Messung (z. B. Zeit-Spannung) plotten
plt.plot(daten[:,0], daten[:,0], "o")  # x=Zeit, y=Spannung

# Achsenbeschriftung und Titel
plt.xlabel("Zeit [s]")
plt.ylabel("Spannung [V]")
plt.title("Versuch 1 verschiedene Abstände")
plt.grid(True)

plt.show()
```
