```python
#Zufallsvariale X = Anrufe zwischen 20-21 Uhr
# poissonverteilt lambda = durchschnitt = 5
# gesucht k = p(x=5)

import math
lam = 5
k= 5
p = (5**5 / math.factorial(k)) * math.exp(-lam)
print("5 kunden rufen an zwischen 20 & 21 P:",p)
```

```python
import math

lambdaW = 5            # durchschnittliche Anrufe pro Stunde
max = 5           # Hotline kann höchstens 5 Anrufe annehmen

# Wahrscheinlichkeit: höchstens 5 Anrufe
pMaxKapazitaet = sum(
    (lambdaW**k / math.factorial(k)) * math.exp(-lambdaW)
    for k in range(max + 1)
)

# Wahrscheinlichkeit: mehr als 5 Anrufe  Überlastung
ueberlastungP = 1 - pMaxKapazitaet

print("P:",ueberlastungP)
```

```python
# a)
# A ist eine stetige Zufallsvariable => Punktwahrscheinlichkeiten sind 0
# P(A = 1/2) = 0
print("a) P(A = 1/2) = 0 ")

# b)
# Wahrscheinlichkeit über die Verteilungsfunktion von A
# F_A(x) = 1/2 * x^3 + 1/2 * x^2   für 0 ≤ x ≤ 1
# P(A ≤ 1/2) = F_A(1/2) = 3/16
print("b) P(A ≤ 1/2) = F_A(1/2) = 3/16")

# c)
# Der Wertebereich von B ist [2, 6]
# Alle möglichen Werte liegen ≤ 6
# P(B ≤ 6) = 1
print("c) P(B ≤ 6) = 1")

# d)
# Über das Gegenereignis rechnen
# P(B > 4) = 1 - P(B ≤ 4)
# F_B(x) = 3/2 - 3/x   für 2 ≤ x ≤ 6
# P(B > 4) = 1 - F_B(4) = 1/4
print("d) P(B > 4) = 1 - F_B(4) = 1/4")
```

```python
# Teilaufgabe 6.2.3 – Erwartungswerte

# Allgemeine Formel für stetige Zufallsvariablen:
# E(X) = Integral x * f_X(x) dx

# ------------------------------------------------
# Erwartungswert E(A)
# Dichte:
# f_A(x) = (3/2)*x^2 + x   für 0 ≤ x ≤ 1

# E(A) = Integral_0^1 x * ((3/2)*x^2 + x) dx
#      = Integral_0^1 ((3/2)*x^3 + x^2) dx
#      = 3/8 + 1/3
#      = 17/24
E_A = 17 / 24                        # Erwartungswert von A
print("E(A)         =", E_A)
# ------------------------------------------------
# Erwartungswert E(B)
# Dichte:
# f_B(x) = 3 / x^2   für 2 ≤ x ≤ 6

# E(B) = Integral_2^6 x * (3 / x^2) dx
#      = Integral_2^6 (3 / x) dx
#      = 3 * ln(x) | 2 bis 6
#      = 3 * ln(3)
E_B = 3 * math.log(3)                # Erwartungswert von B
print("E(B)         =", E_B)
```
