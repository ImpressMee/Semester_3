Vorüberlegungen:
- Steigen oder Fallen die Vogeltreffer über die Jahre?
- Wann passieren die zusammenstöße?
- Welcher Vogel verursacht am meisten schäden?

Datensatz: Vögel und Flugzeug treffer
- Anzahl der Bird Strikes steigt seit 1990 stark an
- Die meisten Zusammenstöße passieren beim Start oder Landeanflug
- Große Vögel verursachen überproportional hohe schäden

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# CSV-Datei einlesen
df = pd.read_csv("birds_and_planes.csv")
```

```python
# Speichern aller relevanten Daten
incident_year = df["Incident Year"]
incident_phase = df["Flight Phase"]
incident_animal = df["Species Name"]
incident_damage = df["Aircraft Damage"]
incident_impact = df["Flight Impact"]
incident_airport = df["Airport"]
```

```python
# Trend seit 1990
year_counts = incident_year.value_counts().sort_index()
plt.title("Jahr und die nummer der Vögel treffer")
plt.plot(year_counts.index, year_counts.values)
plt.xlabel("Year")
plt.ylabel("Number of Bird Strikes")
plt.grid()
plt.show()
print(year_counts)
```

```python
# Flugphase (Start/Landung)
phase_counts = incident_phase.value_counts()
plt.figure()
plt.title("Bird Strikes nach Flugphase")
plt.bar(phase_counts.index, phase_counts.values)
plt.xlabel("Flugphase")
plt.ylabel("Anzahl der Bird Strikes")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
print(phase_counts)
```

```python
# Welche der häufigsten vögel sind betroffen?
bird_count = incident_animal.value_counts().head(10)

plt.figure()
plt.title("Betroffene vögel")
plt.bar(bird_count.index, bird_count.values)
plt.xlabel("Vögel")
plt.ylabel("Anzahl der Bird Strikes")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
print(bird_count)
```

```python
# Welcher Flughafen trifft am meisten vögel?
airport_count = incident_airport.value_counts().head(10)

plt.figure()
plt.title("Flughäfen mit den meisten Vogel treffer")
plt.bar(airport_count.index, airport_count.values)
plt.xlabel("Flughäfen")
plt.ylabel("Anzahl der Bird Strikes")
plt.xticks(rotation=80)
plt.grid(axis="y")
plt.show()
print(bird_count)
```
